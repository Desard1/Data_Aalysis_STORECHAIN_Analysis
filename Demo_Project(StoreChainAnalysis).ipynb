```python
lowest_profit  = df.groupby(["State","City"])["Profit"]\
                            .sum()\
                            .nsmallest(5)

lowest_profit  = lowest_profit.reset_index()

#Here are 5 state city combinations with the lowest profit
```

```python
most_loses = df[df["City"]=="Philadelphia"].groupby("Sub-Category")["Profit"].sum().nsmallest(8).reset_index()


_ = sns.barplot(

    data = most_loses,
    x = "Sub-Category",
    y = "Profit"
    )

plt.axhline(0)
plt.xticks(rotation = 45)
plt.show()
```

```python
only_rows = df[(df["City"]=="Philadelphia") & (df["Profit"]<0)].reset_index(drop=True).drop(columns ="Row ID")

only_rows.describe()


#We can see that the loses are really spread out and not near the mean at all , which suggests some sub-categories are losing much more that others per order
#To try and verify this we can plot a histogram on the data we have
```

```python
_ =  sns.histplot(
    data = only_rows,
    x = "Sub-Category"
)
plt.xticks(rotation = 45)
plt.show()

#7) As we can see from the plot we have to spikes on binders and phones which account for most of the loses , but if we want to see how much impact
# they actually have , we can take those sub-categories out and recalculate the profits

```

```python
without_major_loses = only_rows = df[(df["City"]=="Philadelphia") & (~df["Sub-Category"].isin(["Phones","Binders"]))]\
                                                .reset_index(drop=True)

without_major_loses["Profit"].sum()


#Nope , still cooked , which mean Binders and Phones ARE losing a lot more money than other but not by a Huge margin where it makes the others insignificant


```

```python
#Now lets explore the relationship between discount and profit,I am gonna choose a scatter and fit a line through it 

scatter_data = df.groupby(["Discount","Category"])["Profit"].mean().reset_index().sort_values(by="Profit",ascending=True)
scatter_data.reset_index(drop=True)


_ = sns.lmplot(
    data = scatter_data,
    x = "Discount",
    y = "Profit",
    hue ="Category",
    palette="bright"

)

#As we can see from this plot , the more discount we have on a product , the more money we lose, esspecially in the Tech category , 
# the sweet spot is around 0-20% if we still want to keep the diccount, same thing applies to the other categories 
```

```python
df["Order Date"] = pd.to_datetime(df["Order Date"],format="mixed",errors="coerce")


profit_period = (
    df.groupby(df["Order Date"].dt.to_period("M"))["Profit"]
      .sum()
      .sort_index()
)

profit_period.plot(
    figsize=(12, 4),
    title="Total profit by year and month"
)
plt.ylabel("Profit")
plt.show()

```
